# Controle de fluxo

- Ordem de execução dos programas
- Ajuda o programa tomar decisões
- Regulado por várias declarações condicionais (if, elif, else), loops (for, while), e funções

## Declarações condicionais (if, elif, else)

- Usadas para checar informações antes de executar o código
- Ajuda o programa a tomar decisões
- if executa o código apenas se a condição for **True**
- else  executa o código apenas se a condição for **False**
- Usa comparação e operadores lógicos

```python
# if
a = 7
b = 8

if a < b:
    print(f'{a} é menor que {b}')
```
