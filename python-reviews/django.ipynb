# Django

- Django é um framework de aplicativo Web gratuito e de código aberto escrito em Python.
- Um web framework (estrutura da web) é um conjunto de componentes ou ferramentas que ajudam você a desenvolver sites e aplicativos da web de forma mais rápida e fácil.
- Django não é uma linguagem de programação.

## Criando um App Django

**Django App vs Projects:**

- Um projeto Django é uma coleção de configurações e aplicativos para um site. Ex.: Facebook (como um todo)
- O aplicativo Django é basicamente um aplicativo da web que faz algo. Ex.: Login do Facebook
- Um projeto Django pode conter vários aplicativos, enquanto um aplicativo Django pode ser usado em vários projetos.
- O arquivo `manage.py` é o arquivo Python utilizado para interagir com seu projeto e aplicativos Django.

```bash

python manage.py startapp name-of-app

```
* Iniciar e Adicione um nome ao app

Após iniciar o app django, você deverá abrir o arquivo settings.py da pasta do projeto django e adicionar manualmente o nome do app na lista da variável INSTALLED_APPS.

Como exemplo, criei um app chamado `post`.

## Criando um modelo

Ao abrir a pasta do App `post`você deve selecionar o arquivo `models.py`

- Um modelo é basicamente uma classe Python que você pode usar para criar tabelas em seu banco de dados.

````python
class Post(models.Model):
    title = models.CharField(max_length=200)
    image = models.ImageField(upload_to='images/')
    
    def __str__(self):
        return self.title
````

## Migration

- É uma forma de propagar mudanças para o seu modelo no schema de banco de dados. 
- Eles foram desenvolvidos para funcionar automaticamente, mas precisa saber quando fazer migrações e quando executá-las.
- `makemigrations` é responsável por criar novas migrações baseadas nas mudanças que você fez nos modelos. Depois que a migração for criada, podemos executá-la, o que criará a tabela do banco de dados.

```bash
python manage.py makemigrations
```

* Para criar uma migração

```bash
python manage.py migrate
```
* Para executar a migração

## Definindo rotas de URL

- No arquivo `urls.py` que fica na pasta principal do seu projeto django adicionar o seguinte código:
```python
from django.contrib import admin
from django.urls import path, include
from django.conf import settings
from django.conf.urls.static import static

urlpatterns = [
    path("admin/", admin.site.urls),
    path("", include("post.urls")),
]

if settings.DEBUG:
    urlpatterns += static(settings.MEDIA_URL, document_root=settings.MEDIA_ROOT)
    urlpatterns += static(settings.STATIC_URL, document_root=settings.STATIC_ROOT)

```

Em seguida, crie um arquivo chamado `urls.py` na pasta do seu app e introduza o seguinte código:
```python
from django.urls import path
from .views import HomePageView, CreatePostView # Import the views from the views.py file

urlpatterns = [
    path("", HomePageView.as_view(), name="home"),
    path("post/", CreatePostView.as_view(), name="add_post"),
]
```

## Criando funções view

Crie um arquivo chamado `forms.py` na pasta principal do app e adicione o seguinte código:
```python
from django import forms
from .models import Post

class PostForm(form.ModelForm):
    
    class Meta:
        model = Post
        fields = ['title', 'image']
        
```


No arquivo `views.py", insira o seguinte código:
```python
from django.views.generic import ListView, CreateView
from django.urls import reverse_lazy
from .forms import PostForm
from .models import Post

# Create your views here.
class HomePageView(ListView):
    model = Post
    template_name = 'home.html'
    

class CreatePostView(CreateView):
    model = Post
    form_class = PostForm
    template_name = 'post.html'
    success_url = reverse_lazy('home')

```

## Templates

- Contém duas partes: 
  - Static HTML 
  - Sintaxe especial {% ... %} descrevendo como o conteúdo dinâmico será inserido
- Template é um arquivo de texto que pode gerar qualquer formato baseado em texto (html, xml, csv, etc.)
- Um template contém variáveis e tags.
  - As variáveis ​​são substituídas por valores quando o template é avaliado. 
  - A tag controla a lógica do template.

**Formas de criar template:**
- Pasta Django App > Templates > Django App > home.html
