# Modules

Um módulo Python é basicamente um arquivo Python que pode conter vários tipos de objetos, como funções, variáveis ​​globais, dicionários de lista e outros tipos de objetos.



```python
def say_hello(name):
    """Function to greet a person by name."""
    return f"Hello, {name}!"

fruits = {
    "name": "apple",
    "color": "red",
    "taste": "sweet"
}
```

## Como utilizar um módulo

```python
"""
import module_name
from module_name import function_name
"""
```

```python
import salute

salute.say_hello("Alice") # chamei a função say_hello do módulo salute (craido por mim)
# This is a comment explaining the following line of code
```

```python
from salute import fruits, say_hello
# This imports the fruits dictionary from the salute module
# This imports the say_hello function from the salute module
# Now you can use the function directly

print(fruits["name"])  # Output: apple
print(fruits["color"])  # Output: red
print(fruits["taste"])  # Output: sweet
```

## Pacotes

- São pastas que contém módulos.
- Eles contém um arquivo python especial chamado `__init__.py`
- O arquivo `__init__.py` pode estar vazio. Ele é basicamente usado para informar ao Python que o diretório ou a pasta contém um pacote Python, que pode ser importado como um módulo.
- Os pacotes são uma maneira conveniente de organizar seus módulos e também podem ser aninhados uns nos outros (pacote dentro de um pacote).

```python
import healthy_pack.foo # import pacote.módulo

healthy_pack.foo.fruits("Apple") # chamei a função foo do módulo foo do pacote healthy_pack
```

```python
from healthy_pack import foo # from pacote import módulo

foo.fruits("Apple") # chamei a função foo do módulo foo do pacote healthy_pack
```

## Função dir()

- É uma função interna do Python que você pode usar para listar objetos definidos dentro de um módulo.
- Ele pode ser usado em um módulo personalizado, que é um módulo que você mesmo construiu ou um módulo integrado.

```python
import secrets, uuid # Importing the secrets and uuid modules

s = secrets.token_hex(16)  # Generates a random 16-byte hex token
print(s)  # Prints the random token

t = uuid.uuid4()  # Generates a random UUID
print(t)  # Prints the random UUID

print(" ")  # Prints a separator line

y = dir(secrets)  # Lists all the attributes and methods of the secrets module
print(y)  # Prints the list of attributes and methods

z = dir(uuid)  # Lists all the attributes and methods of the uuid module
print(z)  # Prints the list of attributes and methods
```

## Pycache (.pyc)

- É basicamente uma pasta que contém código Python de três bytes, que já foi compilado e está pronto para ser executado.
- Faz com que seus programas iniciem mais rápido, especialmente quando você está importando um módulo para outro arquivo Python.
- Eles são criados automaticamente após um módulo ser importado para outro arquivo Python e utilizado.
- Eles geralmente são criados no mesmo diretório do próprio módulo.
- O pycache já possui código compilado, o que o torna mais rápido de acessar e executar.

## Atributo Name em Python

- Cada módulo em python tem um avariável especial chamada `name`.
- O valor do atributo name também pode ser definido como o nome do módulo onde o programa é executado.
- O atributo name permite que você execute arquivos Python, seja como módulos reutilizáveis ​​ou como programas autônomos.
- Você usa uma instrução if para verificar se o atributo name está definido como main, que é o valor padrão. Ex.: `if__name__=__main__`
- A declaração `if__name__=__main__` permite executar arquivos python como módulos reutilizáveis ou programas autônomos.
