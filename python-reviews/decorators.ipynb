## Decorators

- É um design python que permite um usuário adicionar uma nova funcionalidade a um objeto existente, seja uma função, um método ou uma classe sem modificar sua estrutura
  - Uma função é um bloco de código reutilizável que pode receber entradas (argumentos), executar operações e retornar um valor.
  - Um método é uma função definida dentro de uma classe e que opera sobre objetos daquela classe. Ele pode acessar e modificar os atributos do objeto.
  - Uma classe é um modelo (ou blueprint=desenho técnico) para criar objetos. Objetos são instâncias de uma classe e contêm atributos (dados) e métodos (funções dentro da classe).
- O decorator é chamado antes da definição de um objeto
- O decorator é representado pelo símbolo @ seguido do nome do objeto decorator

[Decorators for Functions and Methods](https://peps.python.org/pep-0318/)

```python
def my_decorator(function):
    def wrapper():
        myfunc = function()
        convert_uppercase = myfunc.upper()
        return convert_uppercase
    return wrapper

def say_hello():
    return  'hello world'

decorate = my_decorator(say_hello)
print(decorate())

```

```python
def my_decorator(function):
    def wrapper():
        myfunc = function()
        convert_uppercase = myfunc.upper()
        return convert_uppercase
    return wrapper

@my_decorator
def say_hello():
    return  'hello world'
print(say_hello())
```
