# Funções

A finalidade é reutilizar o código em forma de uma função e chamá-la quando necessário

- É uma coleção/bloco de código que pode ser executada em qualquer lugar do seu código
- São reutilizáveis
- São executadas quando chamadas pelo nome atribuído a elas
- Podem ter parâmetros(variáveis) e argumentos(valores)
- Podem retornar dados como resultado
- Existem muitas funções embarcadas no python. Exemplo: print()
- Pode-se criar funções personalizadas

## Criando uma função

```python
def function_name(): # dentro dos parêntesis vão os parâmetros obrigatórios. Se estiver vazio, quer dizer que não tem parâmetro obrigatório
    print('Learning again')

function_name
```

```python
def soma(x,y: float): # parâmetros são as variáveis definidas dentro dos parênteses 
    print(x + y)

soma(4,5) # argumento é o atual valor que você passa para a função, nesse caso, 4 e 5
```

```python
def sum(x,y): # x e y são os parâmentros da função
    value = float(x) + float(y)
    return value

sum(3,4) # Argumento é o valor real passado para a função quando ela é chamada. Esse valor substitui o parâmetro definido na função.
```

## Valor default do parâmetro

- Default: Esse é o valor que uma função usa quando chamada sem passar nenhum valor (argumento)
- Apenas parâmetros no final de uma lista de parâmetros podem ter um valor default 

```python
def divide(x, y=1): # nesse caso, 1 é o valor default para y
    print(x / y)
```

```python
def divide(x=0, y): # nesse caso, não é permitido passar o valor default para o primeiro parâmetro
    print(x / y)
```

```python
def student_names(names='Eudes'): # ** é um parâmetro que aceita um número variável de argumentos, mas ele aceita um dicionário
    print(f'Hello, ', names)

student_names()
```

```python
student_names("João")
```

```python
def student_names(*names): # * é um parâmetro que aceita um número variável de argumentos
    print(f'Hello, ', names)
student_names("João", "Maria", "José")
```

```python
def student_names(**names): # ** é um parâmetro que aceita um número variável de argumentos, mas ele aceita um dicionário
    print(names)

student_names(name1="João", name2="Maria", name3="José")
```

## Argumento de palavras-chave

- Permite especificar quais parâmetros usar de uma lista de parâmetros
- Você não precisa se preocupar com a ordem dos argumentos

```python
def more_num(a, b=7, c=10): # nesse caso, b e c são valores default
    print('a is', a, 'and b is', b, 'and c is', c)
    
more_num(3, 8) # nesse caso, o valor default de c é mantido
more_num(25, c=24) # nesse caso, o valor default de b é mantido
more_num(c=50, a=100) # nesse caso, o valor default de b é mantido
```

## Variável local vs global

- Local: definida dentro de uma função
- Global: definida fora de uma função

```python
x = 10 # global

def my_numbers(x):
    print(x)
    x = 7 # local
    print('My favorite number is', x)
```

```python
my_numbers(x)
```

```python
print(x)
```

```python
x = 10 # global

def my_numbers():
    global x # assinalando a variável como global
    print(x)
    x = 7 # local
    print('My favorite number is', x)
    
my_numbers()
print(x)
```

## Aninhando funções

- Criar uma função dentro da outra


```python
def number(a):
    def num(a):
        return  a + 1
    result = num(a)
    return result

print(number(4))
```

## Parâmetros VarArg

- Uso de * e **
- Quando há um * o argumento é passado como uma tupla
- Quando **, será dicionário

```python
def total_numbers(a=7, *numbers, **phonebook):
    print('My fav number is', a)
    for single_num in numbers: # os argumentos para a tupla serão passados quando a função for chamada
        print('single_num', single_num)
    for name, phonebook_number in phonebook.items(): # quando a função form chamada, os nomes fornecidos serão passados como argumentos para o dicionário
        print(name, phonebook_number)
total_numbers(7, 1, 2, 3, Jack=1123, John=2231, Inge=1560)
```

## Função Anônima

- Lambda
- Podem ter diversos argumentos


```python
a = lambda b: b + 17 # lambda é uma função anônima que pode ter qualquer número de argumentos, mas só uma expressão
print(a(10))
```

```python
c = lambda d,e: d + e
print(c(10, 5))
```

```python
def ghost_number(n):
    return lambda f: f * n
double = ghost_number(2)
print(double(7))
```

## DocStrings (document strings)
- Permite mostrar a documentação do código quando ele é executado
- São acessadas usando dois underlines
- São definidas com aspas triplas

```python
def add_numbers(x, y):
    '''Adding two numbers
    
    The values must be integers.
    '''
    return x + y

add_numbers(3,4)

print(add_numbers.__doc__) # acessando a documentação da função. __doc__ é um atributo que retorna a documentação da função
```
