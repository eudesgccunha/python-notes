# Funções

A finalidade é reutilizar o código em forma de uma função e chamá-la quando necessário

- É uma coleção/bloco de código que pode ser executada em qualquer lugar do seu código
- São reutilizáveis
- São executadas quando chamadas pelo nome atribuído a elas
- Podem ter parâmetros(variáveis) e argumentos(valores)
- Podem retornar dados como resultado
- Existem muitas funções embarcadas no python. Exemplo: print()
- Pode-se criar funções personalizadas

## Criando uma função

```python
def function_name(): # dentro dos parêntesis vão os parâmetros obrigatórios. Se estiver vazio, quer dizer que não tem parâmetro obrigatório
    print('Learning again')

function_name
```

```python
def soma(x,y: float): # parâmetros são as variáveis definidas dentro dos parênteses 
    print(x + y)

soma(4,5) # argumento é o atual valor que você passa para a função, nesse caso, 4 e 5
```

```python
def sum(x,y): # x e y são os parâmentros da função
    value = float(x) + float(y)
    return value

sum(3,4) # Argumento é o valor real passado para a função quando ela é chamada. Esse valor substitui o parâmetro definido na função.
```

## Valor default do parâmetro

- Default: Esse é o valor que uma função usa quando chamada sem passar nenhum valor (argumento)
- Apenas parâmetros no final de uma lista de parâmetros podem ter um valor default 

```python
def divide(x, y=1): # nesse caso, 1 é o valor default para y
    print(x / y)
```

```python
def divide(x=0, y): # nesse caso, não é permitido passar o valor default para o primeiro parâmetro
    print(x / y)
```
