- Erros que ocorrem durante a execução de um programa

- Exceptions inerente: "nome" ou "tipo" do erro. Ex.: NameError, SyntaxError, ValueError
- Handler: "resposta" ou "explicação" do erro

# Handler ('manuseio')

- O Python usará seus próprios métodos integrados para responder a qualquer exceção que ocorra, e isso se dá de maneira genérica.
- É bom definir seus próprios métodos de handler quando seus programas estão sendo construídos.
- Uma das maneiras de fazer isso é incorporar o código que você escreve na instrução **try** e **except**
- O código dentro do bloco **try** é o código que você deseja testar para detectar erros, e o código dentro do bloco **except** conterá o handler de como lidar com esse erro quando ele for detectado.
- Se você não tiver nenhuma instrução try ou except dentro do seu código, o Python usará suas próprias maneiras padrão para lidar com o problema.

- Keywords:
  - try: inclui o bloco de código que deseja testar.
  - except: especifica como lidar com os erros.
  - finaly: especifica qual código será executado independentemente do que acontecer nos blocos try e except.
  - else: especifica qual código executar, mesmo que ocorra um erro de exceção.

```python
# sintax error

try:
    print("Hello, World!")
except SyntaxError as e:
    print("Syntax error:", e)

else:
    print("This block executes if no exception occurs.")
    
finally:
    print("This block always executes; regardless of errors.")
```

```python
try:
    print(x)
except:
    print("Variable not defined")
```

```python
x = 10
try:
    print(x)
except:
    print("Variable not defined")
else:
    print("No exception occurred.")
```

```python
try:
    numero = int(input("Digite um número: "))
    resultado = 10 / numero
except ZeroDivisionError:
    print("Erro: divisão por zero!")
except ValueError:
    print("Erro: entrada inválida! Digite um número válido.")
except Exception as e:
    print(f"Ocorreu um erro: {e}")
```
