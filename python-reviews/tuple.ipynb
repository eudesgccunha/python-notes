# Tuplas

- Uma lista que não pode ser mudada
- Criada usando parêntesis ou um construtor 
- Acessa os elementos por seus index

```python
fruits = ("grapes", "apples", "berries","oranges")
print(fruits[2]) # acessar um item da tupla

print(2*"\n")
print(10*"-")

for fruit in fruits:
    print(fruit)


for items in fruits:
    print(items)
```

```python
# construtor de tuplas
countries = tuple(("Angola", "Brazil", "Portugal", "USA"))
print(countries)
print(countries[1])
```

```python
# tamanho da tupla
print(len(countries))
```

```python
countries.add("new_element") # não é possível adicionar elementos a uma tupla
```
