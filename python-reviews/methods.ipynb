# Método vs Função

- **Função**
  - Uma função é um bloco de código que faz uma uma atividade específica
  - Pode ser passado dados ou parâmetros dentro dos parêntesis
  - Pode criar funções personalizadas
- **Método**
  - Similar a uma função
  - É uma função definida dentro de uma classe e que opera sobre objetos daquela classe. Ele pode acessar e modificar os atributos do objeto
  - Está disponível para um objeto de dado (data type). Ex.: "hello" (string) - nesse exemplo o método string modifica o objeto "hello".
  - Chamados como objeto.método(). Exemplo: "hello".upper()

```python
# uso de métodos
b = "hello"
len(b)
```

```python
b.upper()
```
