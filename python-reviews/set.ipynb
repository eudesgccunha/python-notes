# Set

- Coleção de valores
- Os valores no set não são ordenados 
- Os valores não são indexados
- Você pode adicionar valores ao set, mas não pode mudá-los


- *Métodos*:
    - add() - adiciona um elemento ao set
    - update() - Adiciona multiplos elementos ao set
    - clear() - remove todos os elementos do set
    - del - deleta o set
    - discard() - remove um elemento específico ou um item (tenta remover um elemento, mas se o elemento não existir no conjunto, ele não gera erro)
    - remove() - remove um item específico do set (tenta remover um elemento do conjunto. Se o elemento não existir no conjunto, ele gera um erro do tipo KeyError)
    - union() - combina dois ou mais conjuntos (sets) e retornar um novo conjunto que contém todos os elementos exclusivos de ambos os conjuntos. Ou seja, ele realiza a união dos conjuntos, removendo duplicatas automaticamente, já que um conjunto não pode conter elementos repetidos.



```python
fruits = {"grapes", "apples", "berries"}
print(fruits)
```

```python
for x in fruit:
    print(x)
```

### Set construtor

```python
animals = set(("cat", "dog", "fish")) # set constructor - precisa ser com dois parênteses
print(animals)
```

### Contando valores no Set

```python
print(len(animals)) # contém 3 elementos
```

### Procurando valores no Set

```python
print("cat" in animals) # responde um boolean - True
```

## Métodos

```python
# add() - adiciona um elemento ao conjunto
fruits.add('oranges')
print(fruits)
```

```python
# update() - adiciona vários elementos ao conjunto
fruits.update('tomato', 'pineapple', 'banana') # sem passar como uma lista ele vai adicionar cada letra como um elemento
print(fruits)
```

```python
# update() - adiciona vários elementos ao conjunto
fruits.update(['tomato', 'pineapple', 'banana']) # precisa passar como uma lista
print(fruits)
```

```python
# remove() - remove um elemento do conjunto
fruits.remove("kiwi") # se o elemento não existir ele vai dar um erro
print(fruits)
```

```python
# discard() - remove um elemento do conjunto
fruits.discard("tomato") # se o elemento não existir ele não vai dar erro
print(fruits)

# a diferença entre remove e discard é que o remove vai dar erro se o elemento não existir no conjunto
```

```python
# clear() - remove todos os elementos do conjunto
new_fruits = fruits.copy()
new_fruits.clear()
print(new_fruits)
```

```python
# del - deleta o conjunto inteiro
del new_fruits
print(new_fruits) # vai dar erro pq a variável foi deletada
```

```python
# union() - une dois conjuntos
together = animals.union(fruits)
print(together)
```

```python
# intersection() - interseção entre dois conjuntos
fruits = {"grapes", "apples", "berries"}
intersection = animals.intersection(fruits)
print(intersection) # não tem interseção entre os dois conjuntos, então vai dar um conjunto vazio
```

```python
# difference() - diferença entre dois conjuntos
difference = animals.difference(fruits)
print(difference) # vai mostrar os elementos que estão no animals e não estão no fruits
```
