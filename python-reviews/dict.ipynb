# Dicionário

- É uma coleção de pares de valor-chave
- Chaves são únicas
- Valores podem mudar
- Pode armazenar diferentes tipos de dados (strings e números)

- **Métodos**:
  - get() - retorna o valor de uma chave específica
  - update() - insere um par chave:valor no dicionário especificado
  - clear() - remove todas os pares de chave:valor do dicionário
  - keys() - retorna a lista das chaves do dicionário
  - values() - retorna a lista dos valores do dicionário
  - del - deleta o dicionário
  - pop() - deleta uma chave específica do dicionário 

```python
mycar = {
    "brand": "Range Rover Sport",
    "model": "HSE",
    "year": 2017
}

print(mycar)
```

```python
# função de construção de dicionário
mygreens = dict(fruit="green apples", vegetables='kale')
print(mygreens)
```

## Métodos

```python
# contar a quantidade de valores e chaves
print(len(mycar))
```

```python
# mudar um valor de uma chave
mycar['year'] = 2019 # seleciona a chave e informa o novo valor
print(mycar)
```

```python
# update() 
mycar.update({'color':'silver'})
print(mycar)
```

```python
# get() 
print(mycar.get("brand"))
```

```python
# keys()
mycar.keys()
```

```python
# acessar algum item da lista de chaves
mycar['brand']
```

```python
# values
mycar.values()
```

```python
# pop()
mycar.pop('color')
print(mycar)
```

```python
# clear()
mycar.clear()
print(mycar)
```

```python
del mycar
print(mycar)
```
