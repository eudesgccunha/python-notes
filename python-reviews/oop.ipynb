# Programação Orienta a Objetos (OOP)

- Tipo de programação que qubra coisas em objetos que interagem entre si
- Objetossão criados a partir de classes
- Promove a reutilização do código através do conceito DRY (don't repeat yourself)
- **Objeto** 
  - Um objeto pode ser uma entidade ou qualquer coisa em seu programa que faça algo
- **Classe**
  - É um template para criar objetos
  - Classes podem ter propriedades(*atributos*) que são como **variáveis** que armazenam dados
  - Classes podem ter comportamentos(*métodos*) que funcionam como **funções**.
  - Você pode criar vários objetos a partir de uma única classe
- Instanciar uma classe significa criar um novo objeto a partir de uma classe
- Uma classe pode ser criada com a palavra-chave *class* seguido pelo nome que você quer dar a ela
- A prática comum é fazer com que os nomes de sua classe estejam em PascalCase. Ex: MyCar

```python
class MyInstructors: # nome da classe em PascalCase
    companyName = "Bluelime" # atributo da classe - name da variável em camelCase
    
     # defenindo o método inicial - essa função é usada para inicializar objetos com um valor inicial.
    def __init__(self, course): # Este método é ativado ou chamado automaticamente quando um novo objeto é criado a partir de uma classe.
       
        """
        O parâmetro self é basicamente uma referência à instância atual da classe.
        É o primeiro parâmetro de qualquer método em uma classe e é usado para acessar as variáveis ​​e os métodos pertencentes à classe.
        """
        self.course = course 
    
    
    def print_info(self): # método da classe - name da função em snake_case
        print("My company name is", MyInstructors.companyName)
        print("My course is", self.course) 
```

```python
# você também pode criar uma classe que não tenha métodos

class Pets:
    pass
```

```python
# instanciando uma classe - O processo de criação de um objeto a partir de uma classe é chamado de instanciamento.

class Instructors: # nome da classe em PascalCase
    companyName = "Bluelime" # atributo da classe - name da variável em camelCase
    
     
    def __init__(self, course):       
        self.course = course 

    def print_info(self): 
        print("My company name is", Instructors.companyName)

        
elearning = Instructors("Python for beginners") # instanciando a classe
bls = Instructors("Django for beginners") # instanciando a classe
```

```python
# acessando os métodos da classe (nesse exemplo, a classe só tem um método .print_info())

elearning.print_info()
bls.print_info()
```

```python
# acessando os atributos da classe

elearning.course # não usa parênteses porque é um atributo, não um método
```

```python
print(bls.course) 
```

```python
bls.companyName
```

```python
# modificando uma classe - Você pode adicionar atributos, alterar ou excluir os valores de atributos existentes. Você também pode adicionar métodos ou modificar métodos existentes na classe.

class Instructors: # nome da classe em PascalCase
    companyName = "Bluelime" # atributo da classe - name da variável em camelCase
    
     
    def __init__(self, course, duration):       
        self.course = course 
        self.duration = duration # adicionando um novo atributo

    def print_info(self): 
        print("My company name is", Instructors.companyName)

elearning = Instructors("Python for beginners", 8) 
bls = Instructors("Django for beginners", 5)
print(bls.course)
print(bls.duration)
```

```python
# mudar o valor de um atributo
bls.course = "Django for web development"
bls.duration = 6

print(bls.course)
print(bls.duration)
```

```python
# deletar um atributo

del bls.duration
```

## Classe vs Instância(objeto) variáveis

- **Classe**
  - Definida fora do método
  - Acessada fora da classe pelo *className*
  - Não prefixada
  - Modificações afetam todas as instâncias
  - Indentadas
- **Variáveis instanciadas**
  - Definidas dentro dos métodos
  - Acessada fora da classe pelo *objectName*
  - Prefixada com a palavra *self*
  - Modificações locais
  - Instâncias não são indentadas

## Herança

- Classe pai (superclass ou classe base)
- Classe filho (subclasse ou derivada) - uma classe que herda outra classe

```python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname
    
    def print_name(self):
        print(self.firstname, self.lastname)

florist = Person("Jane", "Smith")
florist.print_name()
```

```python
class Person: # classe pai
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname
    
    def print_name(self):
        print(self.firstname, self.lastname)

florist = Person("Jane", "Smith")
florist.print_name()


class Lawers(Person):
    pass # se criar uma classe sem o __init__ ela herda o __init__ da classe pai
happy_lowers = Lawers("John", "Doe")
happy_lowers.print_name()
```

```python
class Person: # classe pai
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname
    
    def print_name(self):
        print(self.firstname, self.lastname)

florist = Person("Jane", "Smith")
florist.print_name()


class Lawers(Person):
    def __init__(self, fname, lname): # se criar uma classe filha com o __init__ ela sobrescreve o __init__ da classe pai
        self.firstname = fname
        self.lastname = lname
    
    def info(self):
        print(self.firstname, self.lastname)

happy_lowers = Lawers("John", "Doe")
happy_lowers.info()
```

```python
class Person: # classe pai
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname
       
    
    def print_name(self):
        print(self.firstname, self.lastname)

florist = Person("Jane", "Smith")
florist.print_name()


class Lawers(Person):
    def __init__(self, fname, lname):
        Person.__init__(self, fname, lname) # desse modo é possível chamar o __init__ da classe pai
    
    def info(self):
        print(self.firstname, self.lastname)

happy_lowers = Lawers("John", "Doe")
happy_lowers.print_name() # passando o método da classe pai
```

```python
# adicionando atributos para a classe filha

class Person: # classe pai
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname
       
    
    def print_name(self):
        print(self.firstname, self.lastname)

florist = Person("Jane", "Smith")
florist.print_name()


class Lawers(Person):
    def __init__(self, fname, lname, casetype): # adicionando um novo atributo 
        Person.__init__(self, fname, lname) # desse modo é possível chamar o __init__ da classe pai
        self.casetype = casetype
        
    def info(self):
        print(self.firstname, self.lastname, self.casetype)

happy_lowers = Lawers("John", "Doe", "Criminal")
print(happy_lowers.casetype) # passando o método da classe FILHA

happy_lowers.info()
```

## Polimorfismo

- Habilidade de pegar ou ter multiplas e várias formas
- O polimorfismo permite que você defina métodos que estão presentes na classe pai que permite

```python
print(len("hello"))
```

```python
print(len([1, 2, 3, 4]))
```

```python
def addNumbers(a,b,c=1):
    return a + b + c

print(addNumbers(10,9))
print(addNumbers(10,9,3))
```

```python
class UK():
    def capital_city(self):
        print("London is the capital of UK")
        
    def language(self):
        print("English is the primary language")
        
class France():
    def capital_city(self):
        print("Paris is the capital of France")
        
    def language(self):
        print("French is the primary language")
        
class Spain():
    def capital_city(self):
        print("Madrid is the capital of Spain")
    
    def language(self):
        print("Spanish is the primary language")

```

Ambas as funções aqui têm o mesmo tipo de métodos, mas os métodos estão fazendo coisas diferentes. Isso é polimorfismo.

```python
queen = UK()
queen.capital_city()
```

```python
zara = Spain()
zara.capital_city()
```

```python
eilfel = France()
eilfel.language()
```

```python
for country in (queen, zara, eilfel):
    country.capital_city()
    country.language()
```

```python
# acessar o método de outra classe
class UK():
    def capital_city(self):
        print("London is the capital of UK")
        
    def language(self):
        print("English is the primary language")
        
class France():
    def capital_city(self):
        print("Paris is the capital of France")
        
    def language(self):
        print("French is the primary language")
        
class Spain():
    def capital_city(self):
        print("Madrid is the capital of Spain")
    
    def language(self):
        print("Spanish is the primary language")
    
def europe(eu): # passando a classe como argumento
    eu.capital_city() # acessando o método da classe
    eu.language()
    
queen = UK()
zara = Spain()
eilfel = France()

europe(queen)
europe(zara)
europe(eilfel)
```

## Encapsulamento

- Encapsulamento é o processo de restringir o acesso a métodos e variáveis ​​​​em uma classe para evitar a modificação direta dos dados, evitando assim a modificação acidental dos dados.
- Uma classe é um exemplo de encapsulamento porque encapsula os dados que contém de seus métodos e variáveis.
- Métodos e variáveis ​​públicas são acessados ​​de qualquer lugar, já os privados são acessíveis apenas pela própria classe.

Caso você queira marcar os atributos como privado em uma classe, deve inserir duplas underlines antes do atributo.


```python
class Cars:
    def __init__(self, speed, color):
        self.speed = speed
        self.color = color

    def set_speed(self, value):
        self.speed = value

    def get_speed(self):
        return self.speed
    
```

```python
ford = Cars(200, 'red')
nissan = Cars(250, 'blue')
toyota = Cars(180, 'green')
```

```python
ford.set_speed(300)
print(ford.get_speed())
```

```python
# Encapsulation
ford.speed = 400
```

```python
print(ford.get_speed())
```

### Tornando a variável privada

- Com o método print(ford.speed) é possível acessar o atributo speed diretamente. Para evitar isso, é possível usar o conceito de encapsulamento; basta adicionar dois underscores antes do atributo (ex: __speed).
-  Com isso, o atributo não pode ser acessado diretamente.
    - print(ford.__speed) não funcionará

- Para acessar o atributo, é necessário criar um método get e set
    ex:
    def set_speed(self, value):
        self.__speed = value
    
    def get_speed(self):
        return self.__speed

```python
class Cars:
    def __init__(self, speed, color):
        self.__speed = speed # double underscore makes it private
        self.__color = color # double underscore makes it private

    def set_speed(self, value):
        self.__speed = value

    def get_speed(self):
        return self.__speed
    
ford = Cars(200, 'red')
nissan = Cars(250, 'blue')
toyota = Cars(180, 'green')

# Encapsulation
ford.speed = 400
```

Então, agora que temos algum encapsulamento em vigor, seria muito difícil alterar o valor da variável diretamente fora dos métodos na classe.

```python
print(ford.__speed())
```

## Abstração

- Em termos de dados, abstração se refere a fornecer apenas informações essenciais sobre os dados para o mundo externo, ocultando assim os detalhes de segundo plano ou a implementação.
- Então, em poucas palavras, a abstração é usada para esconder os detalhes internos e mostrar apenas as funcionalidades necessárias.
- *Classes abstratas* são classes que contêm um ou mais métodos abstratos.
- Classes abstratas não podem ser instanciadas e requerem subclasses para fornecer implementações para os métodos abstratos.
- Subclasses de uma classe abstrata em Python não precisam implementar os métodos abstratos da classe pai.
- Um método abstrato é um método que é declarado e não contém nenhuma implementação. E requer uma subclasse para sua implementação.

```python
class Shape:
    def area(self):
        pass
    def perimeter(self):
        pass

class Square(Shape):
    def __init__(self, side_length):
        self.side_length = side_length

myshape = Shape()
```

O Python vem com um módulo que fornece a infraestrutura que permite definir classes abstratas: ABC = Abstract Base Classes

```python
from abc import ABC, abstractmethod
# Abstract Base Class (ABC) - Uma classe abstrata é uma classe que não pode ser instanciada e é usada como uma base para outras classes.

class Shape(ABC):
    
    @abstractmethod # decorator para métodos abstratos
    def area(self):
        pass
   
    @abstractmethod # decorator para métodos abstratos
    def perimeter(self):
        pass

class Square(Shape):
    def __init__(self, side_length):
        self.side_length = side_length

myshape = Shape() # Erro: não pode instanciar uma classe abstrata
```

```python
myshape = Square(5) # instanciando a classe filha. Erro: não pode instanciar uma classe abstrata
myshape.area() # Erro: não pode instanciar uma classe abstrata
```

Então, você tem que implementar o método na subclasse para instanciar. Ou seja, precisa fornecer a implementação do método pai dentro da subclasse.

```python
from abc import ABC, abstractmethod
# Abstract Base Class (ABC) - Uma classe abstrata é uma classe que não pode ser instanciada e é usada como uma base para outras classes.

class Shape(ABC):
    
    @abstractmethod # decorator para métodos abstratos
    def area(self):
        pass
   
    @abstractmethod # decorator para métodos abstratos
    def perimeter(self):
        pass

class Square(Shape):
    def __init__(self, side):
        self.__side = side
        
    def area(self):
        return self.__side * self.__side # ou self.__side ** 2
    def perimeter(self):
        return 4 * self.__side 

mysquare = Square(5)
```

```python
print(mysquare.area())
print(mysquare.perimeter())
```
