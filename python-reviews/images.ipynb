# PIL

- O Pill permite que você manipule imagens usando Python.
- Pillow é uma biblioteca Python para processamento de imagens.
- Pillow é basicamente uma versão atualizada do Pill antes de você poder usar pillow para manipular imagens.
- Pode adquirir informações como:
  - formato
  - modo
  - tamanho
  - apresenar a imagem

```python
import PIL

p = dir(PIL)  # Lists all the attributes and methods
print(p)  # Prints the list of attributes and methods
```

```python
from PIL import Image

i = dir(Image)  # Lists all the attributes and methods
print(i)  # Prints the list of attributes and methods
```

```python
# open image
img = Image.open("images/fruits/apple.jpg")

# basic image information
print(img.format)  # JPEG
print(img.size)    # (640, 480)
print(img.mode)    # RGB
print(img.info['dpi'])    # {'dpi': (72, 72), 'jfif': 1, 'jfif_version': (1, 1), 'jfif_density': (72, 72), 'jfif_unit': 1, 'exif': b'Exif\x00\x00II*\x00\x08

# show image
img.show()
```

## Manipular imagens

- Converter em outro formato
- Aplicar escala cinza
- Redimencionar a imagem

```python
# save image
img.save("images/fruits/apple_copy.jpg")  # Save as JPEG
img.save("images/fruits/apple_copy.png")  # Save as PNG
```

```python
# convert image to grayscale
img_gray = img.convert("L")  # Convert to grayscale - L means luminance
img_gray.show()  # Show the grayscale image
img_gray.save("images/fruits/apple_gray.jpg")  # Save as JPEG
```

```python
# banana info
new_img = Image.open("images/fruits/banana.jpg")

print(new_img.format)  # JPEG
print(new_img.size)    # (640, 480)
print(new_img.mode)    # RGB
print(new_img.info['dpi']) 
```

```python
# resize image
img_resized = new_img.resize((320, 240))  # Resize to 320x240 pixels
img_resized.show()  # Show the resized image
```

```python
# rotate image
img_rotated = new_img.rotate(45)  # Rotate by 45 degrees
img_rotated.show()  # Show the rotated image
```

```python
new_img.rotate(45).show() # Show the rotated image without saving it
```

## Criar thumbnails (miniaturas)

Uma miniatura é basicamente uma imagem de visualização compactada da imagem original que é usada como um espaço reservado.

Observações:
- O módulo `glob` é usado para recuperar arquivos e a Path Names Machine, um python especificado.
- O módulo `os`é usado para que o python possa interagir com o sistema operacional

```python
import os, glob
# list all images in a directory
img_dir = "images/fruits/"
img_files = glob.glob(os.path.join(img_dir, "*.jpg"))  # List all JPEG files in the directory
print(img_files)  # Print the list of image files
```

```python
# loop through all images and print their info
for img_file in img_files:
    img = Image.open(img_file)
    print(f"File: {img_file}, Format: {img.format}, Size: {img.size}, Mode: {img.mode}, DPI: {img.info.get('dpi', 'N/A')}")
  # Close the image file after showing it
```

```python
size = 128, 128
for infile in glob.glob("images/fruits/*.jpg"):
    file, ext = os.path.splitext(infile)  # Split the file name and extension
    img = Image.open(infile)
    img.thumbnail(size, Image.LANCZOS)  # Create a thumbnail of the image. LANCZOS minimiza a distorção que você pode obter quando comprimir uma imagem
    img.show()  # Show the thumbnail
    #img.save(file + '.thumbnail', 'JPEG')  # Save the thumbnail
```

## Functions 

```python
def show_image(image_path):
    img = Image.open(image_path)
    img.show()
    return img

def save_image(image, save_path):
    image.save(save_path)
    return save_path

def convert_to_grayscale(image):
    return image.convert("L")  # Convert to grayscale - L means luminance

def get_image_info(image):
    return {
        "format": image.format,
        "size": image.size,
        "mode": image.mode,
        "info": image.info
    }
```
