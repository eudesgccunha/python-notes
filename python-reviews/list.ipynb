```python
# criando uma lista
animals = ["dog", "cat", "bird", "fish", "rabbit"] # opção 1
animals = list(("dog", "cat", "bird", "fish", "rabbit")) # opção 2 - precisa ser com dois parênteses
print(animals)
```

```python
animal = list() # opção 3 - cria uma lista vazia
animal = [] # opção 4 - cria uma lista vazia
print(animal)
```

```python
for x in animals:
    print(x)
```

### Acessando os elementos de uma lista

Index se refere à posição do item na lista

```python
# acessando elemenetos da lista
print(animals[2])
print(animals[-1]) # último elemento (4)
print(animals[1:3]) # do segundo ao terceiro elemento (1 e 2)
print(animals[:3]) # até o terceiro elemento (0, 1 e 2)
print(animals[1:]) # a partir do segundo elemento
```

### Modificando os valores de uma lista


```python
# modificar um valor da lista
animals[3] = "hamster" # basta informar a posição/index e passar o novo valor
print(animals)
```

## Métodos de Lista

```python
# Métodos de lista
fruits = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
vegetables = ["carrot", "lettuce", "onion", "potato", "tomato", "cucumber"]
```

#### .extend() 

```python
# extend() - adiciona elementos de uma lista ao final outra lista (como um JOIN para listas)
fruits.extend(vegetables)
print(fruits)

# ou o inverso
vegetables.extend(fruits)
print(vegetables)
```

#### .append()

```python
# append - adiciona um elemento ao final da lista
fruits.append("grape")
print(fruits)
```

#### .sort()

```python
# sort() - ordena a lista por números e ordem alfabética (ascending or descending)
fruits.sort()
fruits.sort(reverse=True)
print(fruits)
```

#### .count()

```python
# count() - retorna a quantidade de vezes que um valor específico aparece na lista
print(fruits.count("banana"))
```

#### .index()

```python
# index() - retorna o índice da primeira ocorrência do valor especificado
print(fruits.index("apple"))
```

#### .insert()

```python
# insert() - adiciona um elemento em uma posição específica da lista (posição, valor)
fruits.insert(2, "pineapple")
print(fruits)
```

#### .pop()

```python
# pop() - remove o elemento de uma posição específica da lista (posição) - se não informar a posição, remove o último elemento
fruits.pop(2)
print(fruits)
```

#### .remove()

```python
# remove() - remove o primeiro elemento com o valor especificado
vegetables.remove("carrot") # se tiver outro elemento com o mesmo valor, ele não será removido
print(vegetables)
```

#### del

```python
# del() - remove o elemento de uma posição específica da lista (posição)
del fruits[2]
print(fruits)
```

#### .clear()

```python
# clear() - remove todos os elementos da lista
fruits.clear()
print(fruits)
```

#### .reverse()

```python
# reverse() - inverte a ordem dos elementos da lista
vegetables.reverse()
print(vegetables)
```

#### .copy()

```python
# copy() - copia uma lista
new_vegetables = vegetables.copy()
print(new_vegetables)
```

## Lista aninhada

Aninhamento de lista (nested list) ou lista bidimensional.

- É uma coleção de listas; Cada lista é um item.
- Cada lista é uma linha com colunas dentro da coleção;
- As listas são zero-index (começa com 0)


```python
cars = [
    ["Ford", "Volvo", "BMW"],
    ["Toyota", "Honda", "Nissan"],
    ["Audi", "Mercedes", "Volkswagen"],
    ["Fiat", "Renault", "Peugeot"]
]
```

### Acessar elementos em lista aninhada

```python
print(cars[0][1]) # Volvo
print(cars[2][2]) # Volkswagen
```

```python
for row in cars:
    print(row)
```

```python
for row in cars: # o primeiro loop percorre as linhas
    for col in row: # o segundo loop percorre as colunas independentemente do nome das variáveis (row, col, cars)
        print(col)
```
